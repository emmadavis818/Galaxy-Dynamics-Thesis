```python
import galaxykick as Galaxy

import astropy.coordinates as coord
import astropy.units as u
import matplotlib.pyplot as plt
import numpy as np

import gala.dynamics as gd
import gala.potential as gp
from gala.units import galactic
```

```python
noiselevel = 0
w0 = gd.PhaseSpacePosition(pos=[-8.1, 0, 0.02]*u.kpc,vel=[13, 245, 8.]*u.km/u.s)
newpot = Galaxy.MWwNoise(N=noiselevel, units=galactic)
```

```python
w0.xyz
```

```python
neworbit = newpot.integrate_orbit(w0, dt=5 * u.Myr, n_steps=50000)
```

```python
fig, ax = plt.subplots()
neworbit.plot()
plt.show()
```

```python
plt.hist(neworbit.energy(),bins = 100,density = True,color = 'black',alpha=1)
plt.xlabel(f'Energy [{neworbit.energy().unit}]')
plt.show()
```

```python
plt.scatter(neworbit.energy(),neworbit.angular_momentum()[2])
plt.xlabel('energy')
plt.ylabel('z-momentum')
plt.title('The Sun')
plt.show()
```

```python
plt.scatter(neworbit.energy(),neworbit.angular_momentum()[0])
plt.xlabel('energy')
plt.ylabel('x-momentum')
plt.title('The Sun')
plt.show()
```
